## 3. Обработка одного вопроса

Реализуйте функцию `ask_question()`, которая:

1. Отправляет вопрос модели
2. Проверяет, вернула ли модель `function_call` (если решаете использовать `function_call`)
3. Если да — парсит аргументы, вызывает нужный function, возвращает результат модели
4. Получает финальный ответ


```python
import time
from pydantic import BaseModel, Field

# Дополните промпт при необходимости
SYSTEM_PROMPT = """
You are an expert mathematical and logical problem solver.

Rules:
1. Solve carefully and verify the result before finalizing.
2. Return the final answer in the `answer` field only.
3. For boolean questions, output exactly: True or False.
4. For sorting questions, output only the sorted values separated by comma and space.
5. For base-N arithmetic, keep the answer in the same base.
6. Do not include units, commentary, or extra text in the final answer.
"""

class SGR(BaseModel):
    interpretation: str = Field(
        description=(
            "What exactly is being asked. If the statement is ambiguous, choose the most "
            "reasonable interpretation."
        )
    )
    observations: str = Field(
        description="Key facts, constraints, formulas, and hidden assumptions."
    )
    solution: str = Field(
        description="Solve step by step carefully. Check arithmetic and logic."
    )
    verification: str = Field(
        description="Verify the result with a second method or a sanity check."
    )
    reevaluation: str = Field(
        description="If verification fails, briefly fix the solution. Otherwise write 'OK'."
    )
    answer: str = Field(
        description=(
            "Final answer only. "
            "For boolean tasks output True or False. "
            "For sorting tasks output values separated by comma and space. "
            "For base-n arithmetic output only the numeral in that base. "
            "Do not add explanations."
        )
    )


def post_processing(answer: str) -> str:
    s = str(answer).strip()

    if s.lower() in {"yes", "true"}:
        return "True"
    if s.lower() in {"no", "false"}:
        return "False"
    if s.startswith("[") and s.endswith("]"):
        s = s[1:-1].strip()

    return s


def ask_question(question: str) -> str | None:
    """
    Отправляет один вопрос LLM-агенту и возвращает финальный текстовый ответ.

    Returns:
        Текстовый ответ модели (строка) или None при ошибке
    """
    last_error = None

    for attempt in range(3):  # retry, чтобы не терять балл за отсутствие повторных попыток
        try:
            resp = client.chat.completions.parse(
                model=model,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": question},
                ],
                temperature=0.1,
                response_format=SGR,
            )

            msg = resp.choices[0].message
            result: SGR = msg.parsed

            if result is None:
                return None

            return post_processing(result.answer)

        except Exception as e:
            last_error = e
            if attempt < 2:
                time.sleep(1 + attempt)

    print(f"Ошибка: {last_error}")
    return None
```

## 4. Прогон по всему датасету

Запустите агента на всех вопросах из датасета. Для каждого вопроса:
1. Вызовите `ask_question()` для получения ответа
2. Приведите ответ к нужному типу через `parse_answer()`
3. Сформируйте объект `Answer(question_id=..., answer=...)`

> ⏱️ Этот шаг может занять время — в зависимости от провайдера и количества вопросов.  
> При необходимости добавьте `time.sleep()` между запросами для соблюдения rate limits.

```python
from collections import Counter
from tqdm.auto import tqdm

answers: list[Answer] = []
failed: list[dict] = []


def most_frequent(items: list[str]) -> str:
    return Counter(items).most_common(1)[0][0]


N_RUNS = 3

for idx, row in tqdm(dataset.iterrows(), total=len(dataset), desc="Обработка вопросов"):
    question_id = row["question_id"]
    question = row["question"]

    answer_arr: list[str] = []

    # Делаем несколько независимых запусков ask_question().
    for _ in range(N_RUNS):
        answer = ask_question(question)
        if answer is not None:
            answer_arr.append(answer)

    # Если вообще не удалось получить ни одного валидного ответа
    if not answer_arr:
        failed.append({
            "index": idx,
            "question_id": question_id,
            "reason": "no_answer",
        })
        continue

    # Берем наиболее частый ответ
    answer = most_frequent(answer_arr)

    try:
        answers.append(Answer(question_id=question_id, answer=answer))
    except Exception as e:
        failed.append({
            "index": idx,
            "question_id": question_id,
            "reason": f"parse_error: {e}",
            "answer": answer,
            "all_answers": answer_arr,
        })

    # Пауза между запросами (при необходимости)
    # time.sleep(0.5)

print(f"\nГотово!")
print(f"Ответов получено: {len(answers)} из {len(dataset)}")
print(f"Ошибок: {len(failed)}")
```

## 5. Валидация и подсчёт score

Используем функцию `validate_answers()` из `src.utils` для проверки ответов.  
Валидация проверяет:
- **Значение** — ответ должен **точно совпадать** с эталоном (сравнение по строковому представлению)

### Формула score

$$\text{accuracy} = \frac{\text{правильных ответов}}{\text{всего вопросов в датасете}}$$

```python
# Валидация ответов
result = validate_answers(answers, dataset)

total_questions = len(dataset)
total_answered = len(answers)
total_correct = total_answered - len(result["errors"])

# Score считается от общего числа вопросов в датасете
accuracy = total_correct / total_questions if total_questions > 0 else 0

print("=" * 50)
print(f"📊 РЕЗУЛЬТАТЫ")
print("=" * 50)
print(f"Всего вопросов в датасете: {total_questions}")
print(f"Ответов отправлено:        {total_answered}")
print(f"Не удалось ответить:       {len(failed)}")
print(f"Ошибок валидации:          {len(result['errors'])}")
print(f"Правильных ответов:        {total_correct}")
print(f"")
print(f"🎯 Accuracy: {accuracy:.2%} ({total_correct}/{total_questions})")
print("=" * 50)
```

```python
# Детализация ошибок (первые 10)
if result["errors"]:
    print("Примеры ошибок:")
    for err in result["errors"][:10]:
        qid = err["question_id"]
        reason = err["reason"]
        row = dataset[dataset["question_id"] == qid]
        if not row.empty:
            q = row.iloc[0]["question"]
            print(f"\n  [{reason}] {q[:80]}...")
            if reason == "value_mismatch":
                print(f"    Ожидалось: {err['expected']}")
                print(f"    Получено:  {err['actual']}")
else:
    print("✅ Все ответы корректны!")
```

```python
def answer_type(x):
    try:
        val = x
        if isinstance(val, str):
            return "str"
        elif isinstance(val, list):
            return "list"
        elif isinstance(val, int):
            return "int"
        elif isinstance(val, bool):
            return "bool"
        elif isinstance(val, float):
            return "float"
        else:
            return type(val).__name__
    except:
        return "str"

dataset["answer_type"] = dataset["answer"].apply(answer_type)
```

```python
# Accuracy по типам ответов
error_qids = {e["question_id"] for e in result["errors"]}
failed_qids = {f["question_id"] for f in failed}

print("Accuracy по типам ответов:")
print("-" * 40)
for atype in ["int", "float", "bool", "list", "str"]:
    subset = dataset[dataset["answer_type"] == atype]
    total = len(subset)
    wrong = sum(1 for _, r in subset.iterrows()
                if str(r["question_id"]) in error_qids or str(r["question_id"]) in failed_qids)
    correct = total - wrong
    acc = correct / total if total > 0 else 0
    print(f"  {atype:>5}: {acc:.0%}  ({correct}/{total})")
```
